# **Part 1:** **Classical programming activity**


```python
bit_a = 1
bit_b = 0

print("A =", bit_a)
print("B =", bit_b)

print("NOT A =", 1 - bit_a)
print("A AND B =", bit_a & bit_b)
print("A OR B =", bit_a | bit_b)
```
